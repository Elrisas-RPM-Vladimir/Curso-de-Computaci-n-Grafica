```python
# LIBRERÍA: Pygame
# Para qué sirve: crear videojuegos y gráficos 2D
# Se aplica en: videojuegos, simulaciones, animaciones educativas

import pygame
pygame.init()

# Crear la ventana
pantalla = pygame.display.set_mode((500, 350))
pygame.display.set_caption("Ejercicio Pygame")

# FUNCIÓN 1: pantalla.fill() — pintar el fondo de color
pantalla.fill((135, 206, 235))   # color cielo azul
print("Fondo pintado")

# FUNCIÓN 2: pygame.draw.rect() — dibujar rectángulos
pygame.draw.rect(pantalla, (34, 139, 34), (0, 280, 500, 70))   # suelo verde
pygame.draw.rect(pantalla, (220, 50, 50), (60, 230, 40, 50))   # personaje rojo
print("Rectángulos dibujados")

# FUNCIÓN 3: pygame.draw.circle() — dibujar círculos (monedas)
pygame.draw.circle(pantalla, (255, 215, 0), (180, 210), 15)   # moneda 1
pygame.draw.circle(pantalla, (255, 215, 0), (250, 210), 15)   # moneda 2
pygame.draw.circle(pantalla, (255, 215, 0), (320, 210), 15)   # moneda 3
print("Monedas dibujadas")

# FUNCIÓN 4: pygame.font.SysFont() — escribir texto en pantalla
fuente = pygame.font.SysFont("Arial", 26, bold=True)
texto  = fuente.render("Puntos: 300", True, (255, 255, 255))
pantalla.blit(texto, (20, 15))
print("Texto escrito")

# FUNCIÓN 5: pygame.image.save() — guardar captura como imagen PNG
pygame.display.flip()
pygame.image.save(pantalla, "captura_pygame.png")
print("Captura guardada como captura_pygame.png ✅")

pygame.quit()
print("\n✅ Ejercicio Pygame completado")
```
