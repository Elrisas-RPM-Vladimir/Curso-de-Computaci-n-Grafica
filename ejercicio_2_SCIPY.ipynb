```python
# LIBRERÍA: SciPy
# Para qué sirve: hacer cálculos estadísticos y científicos
# Se aplica en: tesis, investigación, análisis de resultados

from scipy import stats
import numpy as np

# Datos simples: notas de dos grupos
grupo_A = [14, 15, 13, 16, 17]  # con tutoría
grupo_B = [10, 11, 12, 10, 13]  # sin tutoría

# FUNCIÓN 1: stats.mean() con numpy — calcular la media
media_A = np.mean(grupo_A)
media_B = np.mean(grupo_B)
print(f"Promedio Grupo A: {media_A}")
print(f"Promedio Grupo B: {media_B}")

# FUNCIÓN 2: stats.describe() — resumen estadístico completo
resumen = stats.describe(grupo_A)
print(f"\nResumen Grupo A:")
print(f"  Cantidad : {resumen.nobs}")
print(f"  Mínimo   : {resumen.minmax[0]}")
print(f"  Máximo   : {resumen.minmax[1]}")
print(f"  Media    : {resumen.mean:.2f}")
print(f"  Varianza : {resumen.variance:.2f}")

# FUNCIÓN 3: stats.shapiro() — prueba si los datos son normales
stat, p = stats.shapiro(grupo_A)
print(f"\nShapiro-Wilk Grupo A: p = {p:.4f}")
if p > 0.05:
    print("Los datos son normales ")
else:
    print("Los datos NO son normales ")

# FUNCIÓN 4: stats.ttest_ind() — compara si hay diferencia real entre grupos
t, p_valor = stats.ttest_ind(grupo_A, grupo_B)
print(f"\nPrueba T: t = {t:.4f}, p = {p_valor:.4f}")
if p_valor < 0.05:
    print("Hay diferencia significativa entre grupos ")
else:
    print("No hay diferencia significativa ")

# FUNCIÓN 5: stats.pearsonr() — correlación entre horas y notas
horas  = [2, 3, 5, 1, 6]
notas  = [10, 11, 13, 10, 16]
r, p   = stats.pearsonr(horas, notas)
print(f"\nCorrelación Pearson: r = {r:.4f}")
print("A más horas de estudio, mayor nota" if r > 0 else "No hay relación clara")

print("\n Ejercicio SciPy completado")
```
